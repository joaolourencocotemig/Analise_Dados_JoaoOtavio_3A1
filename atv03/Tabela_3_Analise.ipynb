```python
bruto = pd.read_excel(path, sheet_name="Sheet1", header=None)
```

```python
bruto
```

```python
# Linhas com dados: Brasil + 5 regiões + 27 UFs (exclui título, cabeçalho e fonte)
dados = bruto.iloc[3:36].copy()
dados = dados.reset_index(drop=True)
dados
```

# Tabela 1: Curso superior (total geral), por região, homens e mulheres

```python
superior = dados[[0, 1, 2, 3]].copy()
superior.columns = [
    "regiao",
    "total",
    "homens",
    "mulheres",
]
```

```python
superior = superior.reset_index(drop=True)

# números não podem ficar como texto
for col in superior.columns[1:]:
    superior[col] = pd.to_numeric(superior[col], errors="coerce")

superior
```

```python
superior.describe().round(2)
```

# Tabela 2: Ciência, Tecnologia, Engenharias e Matemática, por região, homens e mulheres

```python
ciencia_tec = dados[[0, 4, 5, 6]].copy()
ciencia_tec.columns = [
    "regiao",
    "total",
    "homens",
    "mulheres",
]
```

```python
ciencia_tec = ciencia_tec.reset_index(drop=True)

for col in ciencia_tec.columns[1:]:
    ciencia_tec[col] = pd.to_numeric(ciencia_tec[col], errors="coerce")

ciencia_tec
```

```python
ciencia_tec.describe().round(2)
```

# Tabela 3: Educação, Serviços pessoais, Saúde e Bem-estar, por região, homens e mulheres

```python
educacao_saude = dados[[0, 7, 8, 9]].copy()
educacao_saude.columns = [
    "regiao",
    "total",
    "homens",
    "mulheres",
]
```

```python
educacao_saude = educacao_saude.reset_index(drop=True)

for col in educacao_saude.columns[1:]:
    educacao_saude[col] = pd.to_numeric(educacao_saude[col], errors="coerce")

educacao_saude
```

```python
educacao_saude.describe().round(2)
```

## Comparando as estatísticas (describe) das 3 tabelas

Juntando o `describe()` das 3
 tabelas 

```python
comparativo = pd.concat(
    [
        superior.describe().round(2),
        ciencia_tec.describe().round(2),
        educacao_saude.describe().round(2),
    ],
    axis=1,
    keys=["Curso superior (geral)", "Ciência, Tecnologia, Eng. e Matemática", "Educação, Serviços pessoais, Saúde e Bem-estar"],
)
comparativo
```
