```python
# No terminal, uma vez, se precisar: python -m pip install openpyxl pandas
import pandas as pd
import openpyxl
```

```python
path = "CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"

todas = pd.read_excel(path, sheet_name=None, header=None)
print(todas.keys())
```

```python
bruto = pd.read_excel(path, sheet_name="Municípios", header=None)
bruto
```

```python
tabela = bruto.iloc[3:5573].copy()

tabela.columns = [
    "col_vazia",
    "uf",
    "cod_uf",
    "cod_munic",
    "municipio",
    "pop_2010_sinopse",
    "pop_2010_compatibilizada",
    "pop_2022",
]
tabela = tabela.drop(columns=["col_vazia"]).reset_index(drop=True)

for col in ["cod_uf", "cod_munic", "pop_2010_sinopse", "pop_2010_compatibilizada", "pop_2022"]:
    tabela[col] = pd.to_numeric(tabela[col], errors="coerce")

tabela
```

## 5 - Crie uma tabela de população agregada por estado

```python
pop_estado = (
    tabela.groupby("uf")[["pop_2010_compatibilizada", "pop_2022"]]
    .sum()
    .reset_index()
)
pop_estado
```

## 6 - Organize essa tabela por população que mais cresceu entre 2022 e 2010 (Faça a diferença se necessário)

```python
pop_estado["crescimento_2010_2022"] = (
    pop_estado["pop_2022"] - pop_estado["pop_2010_compatibilizada"]
)

pop_estado = pop_estado.sort_values("crescimento_2010_2022", ascending=False).reset_index(drop=True)
pop_estado
```

## 7 - Salve essa tabela nova em csv na pasta do seu projeto

```python
pop_estado.to_csv("populacao_por_estado_crescimento.csv", sep=";", index=False)
print("Arquivo salvo: populacao_por_estado_crescimento.csv")
```

## 8 - Agora trabalhe por município

```python
pop_municipio = (
    tabela.groupby(["uf", "cod_munic", "municipio"])[["pop_2010_compatibilizada", "pop_2022"]]
    .sum()
    .reset_index()
)
pop_municipio
```

## 9 - Organize essa tabela por população que mais cresceu entre 2022 e 2010 (Faça a diferença se necessário)

```python
pop_municipio["crescimento_2010_2022"] = (
    pop_municipio["pop_2022"] - pop_municipio["pop_2010_compatibilizada"]
)

pop_municipio = pop_municipio.sort_values("crescimento_2010_2022", ascending=False).reset_index(drop=True)
pop_municipio
```

## 10 - Salve essa tabela nova em csv na pasta do seu projeto

```python
pop_municipio.to_csv("populacao_por_municipio_crescimento.csv", sep=";", index=False)
print("Arquivo salvo: populacao_por_municipio_crescimento.csv")
```
